```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files
import numpy as np

f = files.upload()
img = cv2.imread(list(f.keys())[0])
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

rows, cols = img.shape[:2]

points1 = np.float32([[50, 50], [200, 50], [50, 200]])
points2 = np.float32([[10, 100], [200, 50], [100, 250]])

matrix = cv2.getAffineTransform(points1, points2)
transformed = cv2.warpAffine(img, matrix, (cols, rows))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(transformed)
plt.title("Affine Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0])
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

rows, cols = img.shape[:2]

points1 = np.float32([[0, 0], [cols-1, 0], [0, rows-1], [cols-1, rows-1]])
points2 = np.float32([[50, 50], [cols-50, 0], [0, rows-50], [cols-1, rows-1]])

matrix = cv2.getPerspectiveTransform(points1, points2)
transformed = cv2.warpPerspective(img, matrix, (cols, rows))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(transformed)
plt.title("Perspective Transformation")
plt.axis("off")

plt.show()
```
